**Experiment 7: Zero-Shot Prompting for Classification and Question Answering**

**Question**

Implement zero-shot prompting for text classification and question answering using a pre-trained language model. Take user input, classify the text without providing examples, and answer a question using a zero-shot prompt.

**Aim**

To apply zero-shot prompting for classification and question-answering tasks using a pre-trained generative AI model.

```python
from transformers import pipeline

print("Libraries loaded successfully!")
```

```python
# Load Zero-Shot Classification Model
classifier = pipeline(
    "zero-shot-classification",
    model="facebook/bart-large-mnli"
)

print("Zero-shot classification model loaded successfully!")
```

```python
# User Input for Classification
text = input("Sayem, Enter a sentence for classification: ")

labels = ["Positive", "Negative", "Neutral"]

result = classifier(
    text,
    candidate_labels=labels
)

print("\n--- Zero-Shot Classification Result ---")
print("Input:", text)
print("Predicted Label:", result["labels"][0])
print("Confidence:", round(result["scores"][0], 3))
```

```python
# Question Answering

from transformers import AutoTokenizer, AutoModelForQuestionAnswering
import torch

qa_model_name = "distilbert-base-cased-distilled-squad"

qa_tokenizer = AutoTokenizer.from_pretrained(qa_model_name)
qa_model = AutoModelForQuestionAnswering.from_pretrained(qa_model_name)

print("Question Answering model loaded successfully!")
```

```python
# Question Answering

context = input("\nSayem, Enter a context: ")
question = input("Sayem, Enter your question: ")

inputs = qa_tokenizer(
    question,
    context,
    return_tensors="pt"
)

with torch.no_grad():
    outputs = qa_model(**inputs)

start = torch.argmax(outputs.start_logits)
end = torch.argmax(outputs.end_logits)

answer = qa_tokenizer.decode(
    inputs["input_ids"][0][start:end+1],
    skip_special_tokens=True
)

confidence = (
    torch.softmax(outputs.start_logits, dim=1)[0][start] *
    torch.softmax(outputs.end_logits, dim=1)[0][end]
).item()

print("\n--- Question Answering Result ---")
print("Question:", question)
print("Answer:", answer)
print("Confidence:", round(confidence, 3))
```

**Observation:**
The zero-shot classification model successfully classified the user-provided sentence without using any task-specific training examples. The QA model generated an answer based on the given context and question.

**Result:**
Zero-shot prompting allows a pre-trained model to perform a new task through instructions and candidate labels without requiring additional fine-tuning for that specific task.
