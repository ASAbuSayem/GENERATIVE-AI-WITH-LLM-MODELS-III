# **Experiment 6: BPE, WordPiece and SentencePiece Tokenization**

Question

Compare BPE, WordPiece, and SentencePiece tokenization strategies using English and technical sentences. Display the generated tokens, count the number of tokens, and analyze how different tokenization methods handle rare and compound words.

Aim

To compare BPE, WordPiece, and SentencePiece tokenization methods and analyze their effect on token count and vocabulary representation.

Brief Theory

Subword Tokenization breaks words into smaller meaningful pieces instead of storing every complete word in the vocabulary.

    BPE (Byte Pair Encoding): Builds subwords by repeatedly merging frequently occurring character/token pairs.

    WordPiece: Selects subwords based on their usefulness in the vocabulary and is commonly used by BERT.
    
    SentencePiece: Performs tokenization directly from raw text and does not require pre-tokenized words. It is commonly used by models such as T5.

```python
from transformers import AutoTokenizer

print("Libraries loaded successfully!")
```

```python
# Load Three Different Tokenizers
# BPE Tokenizer
bpe_tokenizer = AutoTokenizer.from_pretrained("gpt2")

# WordPiece Tokenizer
wordpiece_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")

# SentencePiece Tokenizer
sentencepiece_tokenizer = AutoTokenizer.from_pretrained("t5-small")

print("BPE, WordPiece and SentencePiece tokenizers loaded successfully!")
```

```python
text = input("Sayem, Enter a sentence: ")

print("\nInput Sentence:")
print(text)
```

```python
# Tokenization
# BPE
bpe_tokens = bpe_tokenizer.tokenize(text)

# WordPiece
wordpiece_tokens = wordpiece_tokenizer.tokenize(text)

# SentencePiece
sentencepiece_tokens = sentencepiece_tokenizer.tokenize(text)

# Count tokens
bpe_count = len(bpe_tokens)
wordpiece_count = len(wordpiece_tokens)
sentencepiece_count = len(sentencepiece_tokens)
```

```python
# Display Token Sequences
print("\n========== TOKENIZATION RESULTS ==========")

print("\nBPE Tokenization (GPT-2):")
print(bpe_tokens)
print("Token Count:", bpe_count)

print("\nWordPiece Tokenization (BERT):")
print(wordpiece_tokens)
print("Token Count:", wordpiece_count)

print("\nSentencePiece Tokenization (T5):")
print(sentencepiece_tokens)
print("Token Count:", sentencepiece_count)
```

```python
# Comparison Table
print("\n========== COMPARISON ==========")

print(f"{'Method':<18} {'Model':<20} {'Token Count':<12}")
print("-" * 50)

print(f"{'BPE':<18} {'GPT-2':<20} {bpe_count:<12}")
print(f"{'WordPiece':<18} {'BERT':<20} {wordpiece_count:<12}")
print(f"{'SentencePiece':<18} {'T5-small':<20} {sentencepiece_count:<12}")

print("\nStudent: Abu Sayem")
print("Student ID: 25/SET/MTCE/014")
```
