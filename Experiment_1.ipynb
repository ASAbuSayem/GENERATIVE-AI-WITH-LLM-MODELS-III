Import Library and Load Model

```python
from transformers import pipeline

# Load Pre-trained Generative Model
generator = pipeline(
    "text-generation",
    model="distilgpt2"
)
```

Take User Input and Generate Text

```python
# Take two prompts from the user
prompt1 = input("Sayem, Enter your first prompt: ")
prompt2 = input("Sayem, Enter your second prompt: ")

# Generate text for first prompt
output1 = generator(
    prompt1,
    max_new_tokens=50
)

# Generate text for second prompt
output2 = generator(
    prompt2,
    max_new_tokens=50
)

# Display results
print("\n============================")
print("Experiment 1 - Generative AI")
print("Student: Sayem")
print("Model: DistilGPT-2")
print("Task: Text Generation")
print("==============================")

print("\nFirst Prompt:")
print(prompt1)

print("\nGenerated Output 1:")
print(output1[0]["generated_text"])

print("\nSecond Prompt:")
print(prompt2)

print("\nGenerated Output 2:")
print(output2[0]["generated_text"])
```
