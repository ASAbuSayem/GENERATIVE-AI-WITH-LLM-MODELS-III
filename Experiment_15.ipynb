```python
# Create Streamlit Web Application
%%writefile app.py

import streamlit as st
import torch
import time
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Application configuration
st.set_page_config(
    page_title="LLM Web Application",
    page_icon="🤖",
    layout="centered"
)

# Application title
st.title("🤖 LLM Web Application")
st.write("Generate AI responses using a Hugging Face language model.")

st.markdown("---")

# Student details
st.write("**Student Name:** Sayem")
st.write("**Student ID:** 25/SET/MTCE/014")

# Model configuration
MODEL_NAME = "google/flan-t5-small"

# Load model only once
@st.cache_resource
def load_model():

    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)

    model = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)

    model.eval()

    return tokenizer, model


# Generate response
def generate_response(prompt, tokenizer, model):

    inputs = tokenizer(
        prompt,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    with torch.no_grad():

        outputs = model.generate(
            **inputs,
            max_new_tokens=150,
            do_sample=False
        )

    response = tokenizer.decode(
        outputs[0],
        skip_special_tokens=True
    )

    return response


# Load model
try:

    with st.spinner("Loading language model..."):
        tokenizer, model = load_model()

    st.success("Language model loaded successfully!")

except Exception as e:

    st.error(f"Model loading failed: {e}")
    st.stop()


# User input
st.subheader("Enter Your Prompt")

user_prompt = st.text_area(
    "Write your question or prompt:",
    placeholder="Example: Explain Machine Learning in simple words.",
    height=120,
    max_chars=2000
)

# Generate button
if st.button("Generate Response", type="primary"):

    # Empty input validation
    if not user_prompt.strip():

        st.warning("Please enter a prompt before generating a response.")

    # Input length validation
    elif len(user_prompt.strip()) > 1000:

        st.warning("Please enter a prompt within 1000 characters.")

    else:

        try:

            with st.spinner("Generating response..."):

                start_time = time.time()

                response = generate_response(
                    user_prompt,
                    tokenizer,
                    model
                )

                end_time = time.time()

                generation_time = end_time - start_time

            st.subheader("Generated Response")

            st.success(response)

            st.info(
                f"Response generation time: {generation_time:.2f} seconds"
            )

        except Exception as e:

            st.error(f"An error occurred: {e}")


st.markdown("---")

st.caption("Powered by Streamlit and Hugging Face Transformers")
```
